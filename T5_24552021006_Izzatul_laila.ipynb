## Tugas Praktikum 1: Eksperimen Nilai Ambang Batas (Thresholding)

**Tujuan:**
Memahami pengaruh perubahan nilai ambang batas (*threshold*) terhadap hasil citra biner.

**Instruksi Tugas:**
1. Gunakan citra ***selain `kucing.jpg`***.
2. Lakukan konversi citra tersebut menjadi Grayscale.
3. Buatlah **tiga variasi** nilai ambang batas (*threshold*), yaitu:
   - $T_1 = 80$
   - $T_2 = 128$ (standar)
   - $T_3 = 200$
4. Lakukan proses thresholding untuk masing-masing nilai ambang batas tersebut.
5. Tampilkan hasil ketiga citra biner tersebut secara berdampingan dalam satu baris menggunakan `matplotlib` (gunakan fungsi `plt.subplot`).
6. Tulis analisis singkat di bawah hasil visualisasi mengenai perbedaan yang terlihat pada ketiga gambar hasil thresholding tersebut.

**CONTOH HASIL**
image.png

```python

import cv2
import matplotlib.pyplot as plt
from google.colab import files

# 1. Upload gambar selain kucing.jpg
uploaded = files.upload()

# Ambil nama file gambar yang diupload
nama_file = next(iter(uploaded))

# 2. Membaca gambar
img = cv2.imread(nama_file)

# Periksa apakah gambar berhasil dibaca
if img is None:
    raise ValueError("Gambar tidak berhasil dibaca. Periksa file gambar.")

# 3. Konversi gambar ke Grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 4. Melakukan thresholding dengan tiga nilai ambang batas
T1 = 80
T2 = 128
T3 = 200

_, binary1 = cv2.threshold(gray, T1, 255, cv2.THRESH_BINARY)
_, binary2 = cv2.threshold(gray, T2, 255, cv2.THRESH_BINARY)
_, binary3 = cv2.threshold(gray, T3, 255, cv2.THRESH_BINARY)

# 5. Menampilkan hasil secara berdampingan
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(binary1, cmap='gray')
plt.title('Threshold = 80')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(binary2, cmap='gray')
plt.title('Threshold = 128')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(binary3, cmap='gray')
plt.title('Threshold = 200')
plt.axis('off')

plt.tight_layout()
plt.show()
```

**Analisis Hasil Thresholding**

Dari hasil percobaan, threshold 80 menghasilkan lebih banyak bagian putih, threshold 128 terlihat lebih seimbang, sedangkan threshold 200 menghasilkan lebih banyak bagian hitam. Jadi, semakin tinggi nilai threshold, semakin sedikit bagian putih pada gambar.

### Soal Praktikum 2: Menampilkan Channel Warna Tunggal (Color Channel Isolation)

**Instruksi:**
1. Buatlah program untuk mengisolasi kanal warna **Biru (Blue)** saja dari citra yang sudah diinputkan.
2. Ingat bahwa untuk menampilkan kanal biru murni (bukan grayscale), Anda harus menyetel nilai kanal Merah (R) dan Hijau (G) menjadi `0` pada array citra asli.
3. Tampilkan citra asli berdampingan dengan citra hasil isolasi kanal biru tersebut.

**CONTOH HASIL**
image.png

```python

import cv2
import matplotlib.pyplot as plt

# 1. Membaca gambar yang sudah diupload sebelumnya
img = cv2.imread(nama_file)

# 2. Mengubah gambar dari BGR menjadi RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# 3. Membuat salinan gambar untuk isolasi kanal biru
blue_only = img_rgb.copy()

# 4. Mengatur kanal Merah (R) dan Hijau (G) menjadi 0
blue_only[:, :, 0] = 0
blue_only[:, :, 1] = 0

# 5. Menampilkan gambar asli dan hasil isolasi kanal biru
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title('Citra Asli (RGB)')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(blue_only)
plt.title('Hanya Kanal Biru (B)')
plt.axis('off')

plt.tight_layout()
plt.show()
```

**Analisis Hasil Isolasi Kanal Biru**

Dari hasil percobaan, gambar hanya menampilkan warna biru karena kanal merah dan hijau diubah menjadi 0. Bagian yang memiliki warna biru terlihat lebih jelas, sedangkan bagian lainnya tampak gelap.
