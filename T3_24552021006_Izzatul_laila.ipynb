**Soal 1 (OpenCV & Matplotlib):**
*   Baca gambar `orange.jpg` menggunakan OpenCV, ubah ruang warnanya ke RGB, lalu tampilkan menggunakan Matplotlib dengan ukuran plot $6 \times 6$.

```python
import cv2
import matplotlib.pyplot as plt
import urllib.request

# Download gambar orange.jpg secara otomatis
url = "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/orange.jpg"
urllib.request.urlretrieve(url, "orange.jpg")

# Baca gambar
img = cv2.imread("orange.jpg")

# Ubah BGR menjadi RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Tampilkan gambar
plt.figure(figsize=(6, 6))
plt.imshow(img_rgb)
plt.title("Gambar Orange")
plt.axis("on")
plt.show()
```

**Soal 2 (NumPy - Cropping):**
*   Lakukan pemotongan (cropping) gambar pada koordinat baris/y $200$ hingga $600$ dan kolom/x $400$ hingga $800$ menggunakan slicing array NumPy, lalu tampilkan hasilnya.



```python
# Jawaban Soal No 2: Cropping Gambar
import matplotlib.pyplot as plt

# Memotong gambar menggunakan NumPy
crop = img_rgb[200:600, 400:800]

# Menampilkan gambar hasil cropping
plt.figure(figsize=(6, 6))
plt.imshow(crop)
plt.title('Hasil Cropping Gambar')
plt.axis('on')
plt.show()
```

**Soal 3 (Scikit-Image - Filtering):**
*   Konversi hasil potongan (crop) dari Soal 2 menjadi grayscale, lalu terapkan filter deteksi tepi Sobel (`filters.sobel`) dari Scikit-Image dan tampilkan hasilnya dengan peta warna (cmap) abu-abu (`gray`).

```python
# Jawaban Soal No 3: Deteksi Tepi Sobel
from skimage import color, filters
import matplotlib.pyplot as plt

# Mengubah gambar hasil cropping menjadi grayscale
crop_gray = color.rgb2gray(crop)

# Melakukan deteksi tepi menggunakan filter Sobel
sobel_result = filters.sobel(crop_gray)

# Menampilkan hasil deteksi tepi
plt.figure(figsize=(6, 6))
plt.imshow(sobel_result, cmap='gray')
plt.title('Hasil Deteksi Tepi Sobel')
plt.axis('on')
plt.show()
```
