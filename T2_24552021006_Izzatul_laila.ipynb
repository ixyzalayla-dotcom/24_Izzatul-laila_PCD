### **Tugas 2 — Mengolah String**

Buat program untuk mengolah data nama dan kalimat dengan ketentuan:

- Simpan nama depan dan nama belakang dalam variabel.
- Gabungkan keduanya menjadi nama lengkap.
- Tampilkan sebuah kalimat menggunakan `upper()`, `lower()`, `capitalize()`, dan `title()`.
- Hitung jumlah kemunculan huruf `a` menggunakan `count()`.
- Tampilkan hasil setiap operasi.

Output yang diharapkan: nama lengkap, perubahan format kalimat, dan jumlah karakter.



````python

# Tugas 2: Mengolah String

# Menyimpan nama depan dan nama belakang
nama_depan = "Izzatul"
nama_belakang = "Laila"

# Menggabungkan nama menjadi nama lengkap
nama_lengkap = nama_depan + " " + nama_belakang

print("Nama lengkap:", nama_lengkap)

# Kalimat untuk pengolahan string
kalimat = "saya sedang belajar bahasa python"

# Mengubah seluruh huruf menjadi kapital
print("Uppercase:", kalimat.upper())

# Mengubah seluruh huruf menjadi huruf kecil
print("Lowercase:", kalimat.lower())

# Membuat huruf pertama menjadi kapital
print("Capitalize:", kalimat.capitalize())

# Membuat huruf pertama setiap kata menjadi kapital
print("Title:", kalimat.title())

# Menghitung jumlah huruf a
print("Jumlah huruf a:", kalimat.count("a"))
````

### **Tugas 3 — Operasi Matematika**

Buat program kalkulator sederhana dengan ketentuan:

- Minta pengguna memasukkan dua angka.
- Konversikan input menjadi `int` atau `float`.
- Hitung penjumlahan, pengurangan, perkalian, pembagian, dan pangkat.
- Tampilkan hasil setiap operasi.
- Tambahkan perhitungan luas persegi panjang berdasarkan input panjang dan lebar.

Output yang diharapkan: hasil kalkulasi berdasarkan angka yang dimasukkan pengguna.



````python

# Tugas 3: Operasi Matematika

# Memasukkan dua angka
angka1 = float(input("Masukkan angka pertama: "))
angka2 = float(input("Masukkan angka kedua: "))

# Operasi matematika
print("\nHasil Operasi Matematika")
print("Penjumlahan:", angka1 + angka2)
print("Pengurangan:", angka1 - angka2)
print("Perkalian:", angka1 * angka2)

# Pembagian dengan pengecekan angka nol
if angka2 != 0:
    print("Pembagian:", angka1 / angka2)
else:
    print("Pembagian: tidak bisa dibagi nol")

# Perpangkatan
print("Pangkat:", angka1 ** angka2)

# Menghitung luas persegi panjang
panjang = float(input("\nMasukkan panjang: "))
lebar = float(input("Masukkan lebar: "))

luas = panjang * lebar
print("Luas persegi panjang:", luas)
````

### **Tugas 4 — Perulangan `for` dan `while`**

Buat program untuk menampilkan data mahasiswa dengan ketentuan:

- Buat daftar minimal lima nama mahasiswa.
- Tampilkan semua nama menggunakan `for`.
- Tampilkan angka 1 sampai 10 menggunakan `range()`.
- Tampilkan kembali daftar nama menggunakan `while` dan `len()`.
- Buat perulangan `while` untuk menampilkan bilangan genap dari 2 sampai 20.

Output yang diharapkan: daftar nama dan deret angka yang tampil sesuai ketentuan.


````python

# Tugas 4: Perulangan for dan while

# Membuat daftar lima nama mahasiswa
nama_mahasiswa = [
    "Andi",
    "Budi",
    "Citra",
    "Dina",
    "Eka"
]

# Menampilkan nama menggunakan perulangan for
print("Daftar nama mahasiswa:")
for nama in nama_mahasiswa:
    print(nama)

# Menampilkan angka 1 sampai 10
print("\nAngka 1 sampai 10:")
for angka in range(1, 11):
    print(angka)

# Menampilkan nama menggunakan while dan len()
print("\nDaftar nama menggunakan while:")
i = 0
while i < len(nama_mahasiswa):
    print(nama_mahasiswa[i])
    i += 1

# Menampilkan bilangan genap dari 2 sampai 20
print("\nBilangan genap 2 sampai 20:")
angka = 2
while angka <= 20:
    print(angka)
    angka += 2
````

**Tugas 5 — Fungsi dan Waktu**

Buat program pencatatan aktivitas belajar dengan ketentuan:

- Impor modul `datetime`.
- Buat fungsi bernama `catat_waktu()`.
- Tampilkan pesan `Aktivitas selesai`.
- Tampilkan waktu saat fungsi dijalankan menggunakan `datetime.datetime.now()`.
- Panggil fungsi sebelum dan sesudah perulangan.
- Gunakan perulangan `for` untuk menampilkan proses pengerjaan lima soal.

Output yang diharapkan: pesan aktivitas, waktu pelaksanaan, dan proses pengerjaan soal.




````python

# Tugas 5: Fungsi dan Waktu

# Mengimpor modul datetime
import datetime

# Membuat fungsi untuk mencatat waktu aktivitas
def catat_waktu():
    print("Aktivitas selesai")
    print("Waktu:", datetime.datetime.now())

# Mencatat waktu sebelum pengerjaan
print("Waktu sebelum perulangan:")
catat_waktu()

# Menampilkan proses pengerjaan lima soal
print("\nProses pengerjaan soal:")
for nomor in range(1, 6):
    print("Mengerjakan soal nomor", nomor)

# Mencatat waktu setelah pengerjaan
print("\nWaktu setelah perulangan:")
catat_waktu()
````

## D. Tugas proyek akhir — Sistem Rekap Nilai Mahasiswa

Gabungkan seluruh materi dalam satu program sederhana dengan ketentuan berikut.

Studi kasus

Sebuah program digunakan untuk mencatat data mahasiswa, menerima nilai, menghitung rata-rata, dan menampilkan waktu penyelesaian proses.

Spesifikasi program:

1. Tampilkan judul program menggunakan `print()` dan garis pemisah.
2. Minta input nama dan NIM mahasiswa.
3. Minta pengguna memasukkan nilai tiga mata kuliah.
4. Hitung total dan rata-rata nilai menggunakan operasi matematika.
5. Tampilkan nama mahasiswa dalam format huruf kapital menggunakan `upper()`.
6. Gunakan `for` untuk menampilkan seluruh nilai.
7. Gunakan `while` untuk menampilkan nomor urut mata kuliah.
8. Buat fungsi `tampilkan_hasil()` untuk menampilkan total dan rata-rata.
9. Buat fungsi `catat_waktu()` untuk menampilkan waktu saat program menyelesaikan proses.
10. Tampilkan hasil akhir secara rapi dan mudah dibaca.

### Contoh gambaran output proyek akhir

```
================================
       REKAP NILAI MAHASISWA
================================
Nama       : ANDI PRATAMA
NIM        : 123456
--------------------------------
Nilai Algoritma   : 80
Nilai Basis Data  : 85
Nilai Struktur Data: 90
--------------------------------
Total Nilai : 255
Rata-rata   : 85.0
--------------------------------
Status      : Rekap selesai
Waktu       : [waktu program dijalankan]
================================
```


Data di atas hanya contoh tampilan. Mahasiswa harus membuat program yang menghasilkan nilai berdasarkan input pengguna.


````python

# Proyek Akhir: Sistem Rekap Nilai Mahasiswa

import datetime

# Menampilkan judul program
print("=" * 40)
print("       REKAP NILAI MAHASISWA")
print("=" * 40)

# Meminta data mahasiswa
nama = input("Masukkan nama mahasiswa: ")
nim = input("Masukkan NIM: ")

# Meminta nilai tiga mata kuliah
nilai_algoritma = float(input("Nilai Algoritma: "))
nilai_basis_data = float(input("Nilai Basis Data: "))
nilai_struktur_data = float(input("Nilai Struktur Data: "))

# Menyimpan nilai dalam list
daftar_nilai = [
    nilai_algoritma,
    nilai_basis_data,
    nilai_struktur_data
]

# Menghitung total dan rata-rata
total = sum(daftar_nilai)
rata_rata = total / len(daftar_nilai)

# Fungsi untuk menampilkan hasil nilai
def tampilkan_hasil():
    print("\n" + "-" * 40)
    print("Nama        :", nama.upper())
    print("NIM         :", nim)

    # Menampilkan seluruh nilai menggunakan for
    mata_kuliah = [
        "Algoritma",
        "Basis Data",
        "Struktur Data"
    ]

    print("\nDaftar Nilai:")
    for i in range(len(daftar_nilai)):
        print(mata_kuliah[i], ":", daftar_nilai[i])

    # Menampilkan total dan rata-rata
    print("-" * 40)
    print("Total Nilai :", total)
    print("Rata-rata   :", rata_rata)

    # Menampilkan nomor urut mata kuliah menggunakan while
    print("\nNomor urut mata kuliah:")
    nomor = 1
    while nomor <= len(mata_kuliah):
        print(nomor, ".", mata_kuliah[nomor - 1])
        nomor += 1

    print("Status      : Rekap selesai")

# Fungsi untuk mencatat waktu
def catat_waktu():
    waktu = datetime.datetime.now()
    print("Waktu       :", waktu)

# Menampilkan hasil program
tampilkan_hasil()
catat_waktu()

print("=" * 40)
````
